## The driven pendulum

A rigid pendulum of mass $m$ and length $\ell$ swings in a vertical plane about a pivot. We measure the angle $\theta$ from the hanging position, so $\theta=0$ is the pendulum hanging down and $\theta=\pm\pi$ the pendulum standing upright. A motor at the pivot applies the torque $\tau$, and friction at the pivot slows the motion in proportion to the angular velocity $\omega=\dot\theta$. Newton's law for the rotation, $m\ell^2\ddot\theta=-mg\ell\sin\theta-m\ell^2 b\,\dot\theta+\tau$, divided by $m\ell^2$, reads

$$ \ddot\theta=-\omega_0^2\sin\theta-b\,\dot\theta+u,\qquad \omega_0^2=\frac g\ell,\qquad u=\frac{\tau}{m\ell^2},\qquad |u|\le u_{\max}, \tag{1}$$

with the natural angular frequency $\omega_0$ of small oscillations, the damping rate $b$ and the control $u$, the torque per moment of inertia. We measure time in units of $1/\omega_0$, so $\omega_0=1$, the period of small oscillations is $2\pi$, and $u$ is the motor torque in units of the largest gravitational torque $mg\ell$. Throughout the notebook $b=0.1$, a weakly damped pendulum whose small-angle amplitude decays by the factor $e^{-b t/2}$ without drive, and $u_{\max}=0.3$.

The energy per moment of inertia, measured from the hanging position at rest,

$$ E(\theta,\omega)=\tfrac12\,\omega^2+\omega_0^2\,(1-\cos\theta), \tag{2}$$

is conserved without friction and drive. Along a solution of equation (1) it changes at the rate

$$ \frac{dE}{dt}=\omega\,\dot\omega+\omega_0^2\sin\theta\,\dot\theta=\omega\,u-b\,\omega^2. \tag{3}$$

The motor adds energy when the torque has the sign of the angular velocity and removes it when the signs differ. The upright pendulum at rest has the energy $E_{\rm s}=2\omega_0^2$, and to reach the top the motor has to raise the energy from $0$ to $E_{\rm s}$.

A positive constant torque shifts the stable equilibrium near the bottom to $\theta^*=\arcsin(u/\omega_0^2)$. Its transient turning point follows from the work done by the motor. Starting from rest at $\theta=0$ without damping, integration of equation (3) gives $E=u\theta$. At the first nonzero turning point, $\omega=0$, and therefore

$$ \omega_0^2(1-\cos\theta_{\rm turn})=u\theta_{\rm turn}. $$

For the chosen torque, this equation has a nonzero root below the upright angle. Damping removes energy and lowers the first excursion. The next cell checks energy conservation without drive and damping, then compares constant-torque trajectories with the nonlinear turning-point equation. The integrator uses five fourth-order Runge–Kutta substeps over each decision interval $\Delta t=0.1$.

```python
W0, B, UMAX = 1.0, 0.1, 0.3      # omega_0 (unit of frequency), damping rate b, torque limit u_max
DT, NSUB = 0.1, 5                 # time between two decisions, Runge-Kutta substeps per decision
E_SEP = 2 * W0**2                 # energy of the upright pendulum at rest, the separatrix

def rhs(x, u, b=B):
    # equation (1) as a first-order system for x = (theta, omega)
    th, om = x[..., 0], x[..., 1]
    return jnp.stack([om, -W0**2 * jnp.sin(th) - b * om + u], -1)

def propagate(x, u, b=B):
    # fourth-order Runge-Kutta over one interval DT with the torque u held constant
    h = DT / NSUB
    def body(x, _):
        k1 = rhs(x, u, b); k2 = rhs(x + h / 2 * k1, u, b); k3 = rhs(x + h / 2 * k2, u, b); k4 = rhs(x + h * k3, u, b)
        return x + h / 6 * (k1 + 2 * k2 + 2 * k3 + k4), None
    return lax.scan(body, x, None, NSUB)[0]

def energy(x):
    # equation (2)
    return 0.5 * x[..., 1]**2 + W0**2 * (1 - jnp.cos(x[..., 0]))

@partial(jax.jit, static_argnums=(2,))
def free_run(x0, u, n, b=B):
    # n intervals with a constant torque u; returns the n+1 states
    xs = lax.scan(lambda x, _: (propagate(x, u, b),) * 2, x0, None, n)[1]
    return jnp.concatenate([x0[None], xs])

xs_free = free_run(jnp.array([2.5, 0.0]), 0.0, 1000, 0.0)
E_drift = float(jnp.abs(energy(xs_free) - energy(xs_free[0])).max())
xs_const = np.asarray(free_run(jnp.array([0.0, 0.0]), UMAX, 400))
th_max_const = float(xs_const[:, 0].max())
# Nonzero undamped turning point from motor work and pendulum energy.
lo,hi=1e-8,np.pi
for _ in range(80):
    mid=(lo+hi)/2
    if W0**2*2*np.sin(mid/2)**2-UMAX*mid>0:hi=mid
    else:lo=mid
theta_turn=(lo+hi)/2
xs_undamped=np.asarray(free_run(jnp.array([0.0,0.0]),UMAX,400,0.0))
assert abs(W0**2*(1-np.cos(theta_turn))-UMAX*theta_turn)<1e-12
assert th_max_const<theta_turn
assert abs(xs_undamped[:,0].max()-theta_turn)<1e-3
print(f"no friction, no torque, 100 time units from theta = 2.5: largest change of the energy {E_drift:.1e}")
print(f"constant torque u_max from rest: largest angle {th_max_const:.3f} rad = {np.degrees(th_max_const):.1f} deg, "
      f"nonlinear undamped turning point = {theta_turn:.6f} rad, upright = pi")
NUMBERS.update(Wzero=W0, Damping=B, Umax=UMAX, Dt=DT, Esep=E_SEP, EDrift=f"{E_drift:.0e}",
               ThMaxConstDeg=round(np.degrees(th_max_const), 1), ThBoundConstDeg=round(np.degrees(theta_turn), 1))
```

The energy of the free pendulum changes by less than $10^{-8}$ over one hundred time units, so the integrator conserves what it should. With the full torque pushing towards the goal the pendulum rises to $32.7$ degrees, below the nonlinear undamped turning point, $\theta_{\rm turn}=0.619567$ radians, about $35.5$ degrees, and then oscillates about the shifted equilibrium while friction damps the oscillation. The top lies at 180 degrees.

The constant-torque motion (orange) stays in a small region near the bottom. To reach the top the controller has to carry the state outwards across the closed orbits up to the separatrix, and equation (3) says how, by pushing along $\omega$.

The classical solution of control theory uses equation (3) directly. Åström and Furuta [K. J. Åström and K. Furuta, Automatica **36**, 287 (2000), [doi:10.1016/S0005-1098(99)00140-5](https://doi.org/10.1016/S0005-1098(99)00140-5)] control the energy of a pendulum whose pivot is accelerated, and the same idea applied to a torque at the pivot gives the rule

$$ u=u_{\max}\,\operatorname{sign}\!\big[\omega\,(E_{\rm s}-E)\big], \tag{4}$$

which pushes along the motion while the energy is too low and against it while the energy is too high. At rest at the bottom $\omega=0$, and we then choose $u=+u_{\max}$ to start the motion. Rule (4) brings the energy to the separatrix, and the pendulum then approaches the top along the separatrix. Balancing the pendulum at the top is, in the work of Åström and Furuta, the task of a second, linear controller. We use rule (4) as the physical reference for what the learned policies do.

```python
def energy_controller(x):
    # rule (4), with +u_max when omega = 0
    s = jnp.where(x[1] * (E_SEP - energy(x)) >= 0, 2, 0)
    return jnp.where(jnp.abs(E_SEP - energy(x)) < 1e-12, 1, s)
```

## The swing-up as a Markov decision process

A Markov decision process consists of states $s$, actions $a$, rewards $r$ and the dynamics $p(s',r\mid s,a)$, the probability that action $a$ in state $s$ leads to the state $s'$ with the reward $r$. The swing-up fits this frame once we fix four choices.

The state is the point $(\theta,\omega)$ of the phase plane, which by equation (1) determines the future motion for a given torque, so the process is Markov. The agent decides every $\Delta t=0.1$, a tenth of a radian of the small-oscillation phase, and holds the chosen torque until the next decision. The actions are the three torques $u\in\{-u_{\max},0,+u_{\max}\}$, labelled $a=0,1,2$. Three values suffice for a strategy of the type of rule (4) below, which uses only the extreme torques. The reward after each interval is the height of the bob above the pivot in units of $\ell$,

$$ r_{t+1}=-\cos\theta_{t+1}\in[-1,1], \tag{5}$$

which is $-1$ hanging and $+1$ upright. This reward is the one of Doya's swing-up, with the angle measured from the hanging position, and it says nothing about energy or pumping. Any strategy has to be found by the agent. The return $G_t=\sum_{k\ge0}\gamma^k r_{t+k+1}$ weighs a reward $k$ intervals ahead by $\gamma^k$. We take $\gamma=0.98$, so rewards about $1/(1-\gamma)=50$ intervals ahead, five time units, still count with weight $e^{-1}$. This horizon is shorter than one period $2\pi$ of the small oscillation, and we will vary $\gamma$ to see how much look-ahead the swing-up needs. Episodes last 30 time units, 300 decisions.

The phase plane is a continuum, and value iteration and tabular Q-learning need a finite set of states. We cover the region $-\pi\le\theta<\pi$, $|\omega|\le3$ with a grid of $51\times51$ points. The angle is periodic, so the grid wraps around in $\theta$, and the grid points lie at $\theta_i=-\pi+2\pi(i+\tfrac12)/51$, which puts $\theta=0$ on a grid point and the upright position halfway between two. The velocity grid $\omega_j=-3+6j/50$ covers the separatrix, whose largest velocity is $2\omega_0$, with a margin. Each grid point is one state $s$, $2601$ states in total.

The transition probabilities come from the physics. From the grid point $s=(\theta_i,\omega_j)$ we integrate equation (1) over $\Delta t$ with the torque of action $a$ and land at a point $x'$ between four grid points. We assign the four neighbours the bilinear interpolation weights of $x'$ as probabilities,

$$ p(s'\mid s,a)=w_{s'}(x'),\qquad \sum_{s'}w_{s'}(x')=1, \tag{6}$$

The stored angular coordinates wrap at the periodic boundary. To interpret their weighted mean, lift neighbouring angles by multiples of $2\pi$ into the local interval containing the endpoint. In that locally unwrapped chart, $\sum_jw_j\widetilde\theta_j=\widetilde\theta^{\prime}$, and $\sum_jw_j\omega_j=\omega^{\prime}$ when velocity clipping is inactive. For a periodic function $f$, interpolation uses $\sum_jw_jf(\theta_j,\omega_j)$ without a global Euclidean-coordinate interpretation. The reward of the transition is the height (5) at the end point $x'$, a deterministic function $r(s,a)$. The discretisation turns the deterministic pendulum into a finite Markov decision process with stochastic transitions, whose randomness models the uncertainty of where in the cell the pendulum is. Grids with interpolation weights read as transition probabilities are a standard way to build a finite model of a continuous control problem [R. Munos and A. Moore, Mach. Learn. **49**, 291 (2002), [doi:10.1023/A:1017992615625](https://doi.org/10.1023/A:1017992615625)]. The velocity of an end point outside $|\omega|\le3$ is clipped to the edge of the grid, which never matters for the motions below.

```python
GAMMA = 0.98
ACTS = jnp.array([-UMAX, 0.0, UMAX])          # torques of the actions a = 0, 1, 2
NT, NW, WMAX = 51, 51, 3.0
TH_GRID = -jnp.pi + 2 * jnp.pi * (jnp.arange(NT) + 0.5) / NT
W_GRID = jnp.linspace(-WMAX, WMAX, NW)
XS = jnp.stack(jnp.meshgrid(TH_GRID, W_GRID, indexing="ij"), -1).reshape(-1, 2)   # state s = i * NW + j
S = XS.shape[0]

def wrap(th):
    return (th + jnp.pi) % (2 * jnp.pi) - jnp.pi

def bilinear(x):
    # the four grid points around x and their bilinear weights, equation (6); theta wraps around
    ti = (wrap(x[..., 0]) + jnp.pi) / (2 * jnp.pi) * NT - 0.5
    i0 = jnp.floor(ti).astype(int); ft = ti - i0
    wi = jnp.clip((x[..., 1] + WMAX) / (2 * WMAX) * (NW - 1), 0, NW - 1)
    j0 = jnp.clip(jnp.floor(wi).astype(int), 0, NW - 2); fw = wi - j0
    i0m, i1 = i0 % NT, (i0 + 1) % NT
    idx = jnp.stack([i0m * NW + j0, i1 * NW + j0, i0m * NW + j0 + 1, i1 * NW + j0 + 1], -1)
    wts = jnp.stack([(1 - ft) * (1 - fw), ft * (1 - fw), (1 - ft) * fw, ft * fw], -1)
    return idx, wts

def cell(x):
    # nearest grid point of a phase-space point, the state that the agent sees
    i = jnp.round((wrap(x[..., 0]) + jnp.pi) / (2 * jnp.pi) * NT - 0.5).astype(int) % NT
    j = jnp.clip(jnp.round((x[..., 1] + WMAX) / (2 * WMAX) * (NW - 1)).astype(int), 0, NW - 1)
    return i * NW + j

X_END = jnp.stack([propagate(XS, u) for u in ACTS], 1)           # end points, shape (S, 3, 2)
IDX, WTS = bilinear(X_END)                                         # p(s'|s,a) as 4 neighbours and weights
R = -jnp.cos(X_END[..., 0])                                         # reward r(s,a), equation (5)

dth = jnp.sum(WTS * wrap(XS[IDX][..., 0] - X_END[..., 0][..., None]), -1)   # mean angle offset, wrapped
xe_mean_w = jnp.sum(WTS * XS[IDX][..., 1], -1)
inner = jnp.abs(X_END[..., 1]) < WMAX
S_HANG = int(cell(jnp.array([0.0, 0.0])))
print(f"{S} states, 3 actions, 4 successors each; rows of p sum to one within {float(jnp.abs(WTS.sum(-1) - 1).max()):.1e}")
print(f"largest local-chart coordinate error with velocity clipping inactive: "
      f"{float(jnp.where(inner, jnp.hypot(dth, xe_mean_w - X_END[..., 1]), 0).max()):.1e}")
print("hanging state: grid point", np.asarray(XS[S_HANG]))
NUMBERS.update(Gamma=GAMMA, NGrid=NT, NStates=S)
# A seam endpoint needs locally lifted angles, while its stored arithmetic mean is zero.
seam=jnp.array([jnp.pi,0.0]);seam_idx,seam_w=bilinear(seam)
stored=XS[seam_idx,0]
lifted=seam[0]+wrap(stored-seam[0])
assert np.isclose(float(jnp.sum(seam_w*lifted)),np.pi)
assert np.isclose(float(jnp.sum(seam_w*stored)),0.0)
print('Periodic-seam local-chart test passes')
```

The rows of the transition matrix sum to one. The code checks coordinate reproduction in a locally unwrapped angular chart, with velocity clipping excluded. An arithmetic mean of the stored angles would fail across the periodic seam. The figure shows the construction for one grid point. The three arrows are the motions over $\Delta t$ under the three torques, and the circles on the grid points around the end point of $+u_{\max}$ have areas proportional to the probabilities $p(s'\mid s,a)$.

## Value iteration with the model

The optimal value $V_*(s)$ is the largest expected return from $s$ over all policies, and it obeys the Bellman optimality equation

$$ V_*(s)=\max_a\Big[r(s,a)+\gamma\sum_{s'}p(s'\mid s,a)\,V_*(s')\Big],\qquad Q_*(s,a)=r(s,a)+\gamma\sum_{s'}p(s'\mid s,a)\,V_*(s'). \tag{7}$$

Value iteration applies the right-hand side $\mathcal T$ repeatedly, $V_{k+1}=\mathcal TV_k$ from $V_0=0$. The map $\mathcal T$ contracts distances in the maximum norm by at least $\gamma$, so $\|V_k-V_*\|_\infty\le\gamma^k\|V_0-V_*\|_\infty$ [Sutton and Barto, Sec. 4.4]. With four successors per state and action, one sweep over all $2601$ states is a few gathers of arrays. The optimal policy picks the best action in every state, $\pi_*(s)=\arg\max_aQ_*(s,a)$. The next cell runs value iteration far beyond convergence to obtain the reference $V_*$ and then records the error of every iterate.

```python
def q_from_v(V, gamma):
    # right-hand side of equation (7) for all states and actions
    return R + gamma * jnp.sum(WTS * V[IDX], -1)

@partial(jax.jit, static_argnums=(1,))
def value_iteration(gamma, n):
    return lax.fori_loop(0, n, lambda k, V: q_from_v(V, gamma).max(1), jnp.zeros(S))

@partial(jax.jit, static_argnums=(2,))
def vi_errors(gamma, V_ref, n):
    def body(V, _):
        Vn = q_from_v(V, gamma).max(1); return Vn, jnp.abs(Vn - V_ref).max()
    return lax.scan(body, jnp.zeros(S), None, n)[1]

t0 = time.time()
V_star = value_iteration(GAMMA, 4000)
Q_star = q_from_v(V_star, GAMMA)
PI_STAR = Q_star.argmax(1)
errs = np.asarray(vi_errors(GAMMA, V_star, 1500))
k_ = np.arange(1, len(errs) + 1)
rate = (errs[1000] / errs[500])**(1 / 500)
print(f"value iteration, 4000 sweeps: {time.time() - t0:.1f} s; Bellman residual {float(jnp.abs(Q_star.max(1) - V_star).max()):.1e}")
print(f"|V_0 - V_*| = {float(jnp.abs(V_star).max()):.2f}; error after 100, 500, 1000 sweeps: "
      f"{errs[99]:.2e}, {errs[499]:.2e}, {errs[999]:.2e}; measured rate per sweep {rate:.4f}, gamma = {GAMMA}")
print(f"V_* at the hanging state {float(V_star[S_HANG]):.2f}, largest V_* {float(V_star.max()):.2f}, "
      f"upright forever would give 1/(1-gamma) = {1/(1-GAMMA):.0f}")
NUMBERS.update(ViRate=round(float(rate), 4), VstarHang=round(float(V_star[S_HANG]), 2), VstarMax=round(float(V_star.max()), 2))

fig, ax = plt.subplots(figsize=(5.2, 2.6))
ax.semilogy(k_, errs, color=UJBlue, label=r"$\|V_k-V_*\|_\infty$")
ax.semilogy(k_, GAMMA**k_ * float(jnp.abs(V_star).max()), color=DSOrange, ls="--", label=r"$\gamma^k\,\|V_0-V_*\|_\infty$")
ax.set_xlabel(r"sweep $k$"); ax.set_ylabel("error of the value"); ax.set_ylim(1e-12, 1e3); ax.legend()
fig.tight_layout(); save(fig, "fig_lab056_vi_convergence"); plt.show()
```

The error of value iteration (blue) lies on the contraction bound $\gamma^k\|V_0-V_*\|_\infty$ (dashed) and decreases by the factor $0.980$ per sweep, the discount factor itself. The bound is reached because of the upright state. A policy that holds the pendulum upright collects a reward close to one at every step, and the $k$-th iterate there is the partial sum $\sum_{n<k}\gamma^n$, whose distance to the full sum $1/(1-\gamma)=50$ is $\gamma^k/(1-\gamma)$. The largest optimal value is $48.74$, close to this limit, and from the hanging position $V_*=-32.10$, since the pendulum spends the first part of every episode below the pivot, where the reward is negative. On a grid of a few thousand states value iteration converges to round-off in about a second.

The value itself is the expected discounted height, and its numbers are hard to read. What matters is the policy, the torque that $\pi_*$ chooses at each point of the phase plane. The map below colours every grid point by its optimal action.

```python
@partial(jax.jit, static_argnums=(2,))
def rollout_table(table, x0, n):
    # the continuous pendulum (1) driven by a tabular policy; the agent sees the nearest grid point
    def body(x, _):
        a = table[cell(x)]; return propagate(x, ACTS[a]), (x, a)
    xT, (xs, acts) = lax.scan(body, x0, None, n)
    return jnp.concatenate([xs, xT[None]]), acts

@partial(jax.jit, static_argnums=(1,))
def rollout_rule(x0, n):
    # the continuous pendulum driven by the energy rule (4)
    def body(x, _):
        a = energy_controller(x); return propagate(x, ACTS[a]), (x, a)
    xT, (xs, acts) = lax.scan(body, x0, None, n)
    return jnp.concatenate([xs, xT[None]]), acts

def upright(xs):
    # within 45 degrees of the upright position
    return np.cos(np.asarray(xs)[:, 0]) < -np.cos(np.pi / 4)

def summary(xs):
    # first time within 45 degrees of upright, time from which it stays there until the end,
    # fraction of the last 10 time units upright, largest height
    up = upright(xs); first = np.argmax(up) * DT if up.any() else np.inf
    stay = (len(up) - np.argmax(~up[::-1])) * DT if up[-1] else np.inf
    return first, stay, up[-100:].mean(), float(np.max(-np.cos(np.asarray(xs)[:, 0])))

N_EVAL = 400                                   # 40 time units
X_HANG = jnp.array([0.0, 0.0])
traj_opt, act_opt = rollout_table(PI_STAR, X_HANG, N_EVAL)
PI_GREEDY = R.argmax(1)                        # gamma = 0, the largest immediate reward
traj_gr, act_gr = rollout_table(PI_GREEDY, X_HANG, N_EVAL)
traj_en, act_en = rollout_rule(X_HANG, N_EVAL)
for nm, tr in (("optimal", traj_opt), ("greedy", traj_gr), ("energy rule (4)", traj_en)):
    f, stay, st, hmax = summary(tr)
    print(f"{nm:16s} first within 45 deg of upright at t = {f:5.1f}, stays there from t = {stay:5.1f}, "
          f"fraction of the last 10 time units upright {st:.2f}, largest height {hmax:.3f}")
om_ = np.asarray(traj_opt[:-1, 1]); f_opt = summary(traj_opt)[1]
pre = np.arange(len(om_)) * DT < f_opt
n_rev = int(np.sum(np.diff(np.sign(om_[pre])) != 0))
print(f"optimal swing-up: {n_rev} reversals of the motion before the top")
NUMBERS.update(TfirstOpt=round(summary(traj_opt)[0], 1), TupOpt=round(f_opt, 1), TfirstEnergy=round(summary(traj_en)[0], 1),
               StayEnergy=round(summary(traj_en)[2], 2), HmaxGreedy=round(summary(traj_gr)[3], 2), NRevOpt=n_rev,
               ThMaxGreedyDeg=round(float(np.degrees(np.abs(np.asarray(traj_gr[:, 0])).max())), 1))

fig, ax = plt.subplots(figsize=(5.2, 3.2))
draw_policy(ax, PI_STAR, traj_opt[:int(f_opt / DT) + 30])
fig.tight_layout(); save(fig, "fig_lab056_policy"); plt.show()
```

## Greedy against optimal

## Q-learning without the model

Value iteration read the arrays of the model, the end points of every grid point under every torque. A real controller only observes what happens when it acts. Q-learning [C. J. C. H. Watkins and P. Dayan, Mach. Learn. **8**, 279 (1992), [doi:10.1007/BF00992698](https://doi.org/10.1007/BF00992698)] learns the action values from such observations, one transition $(s,a,r,s')$ at a time, by moving the estimate towards a sampled version of the right-hand side of equation (7),

$$ Q(s,a)\ \leftarrow\ Q(s,a)+\alpha\Big[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\Big]. \tag{8}$$

The agent now interacts with the continuous pendulum (1). It sees only the nearest grid point $s$ of the true state, chooses a torque, the pendulum moves for $\Delta t$, and the agent receives the height (5) and the nearest grid point $s'$ of the new state. It never sees the transition probabilities (6). Actions follow the $\varepsilon$-greedy rule, a random torque with probability $\varepsilon=0.1$ and the torque of largest $Q(s,a)$ otherwise. Each episode of 300 decisions starts at a random point with $\theta$ uniform in $[-\pi,\pi)$ and $\omega$ uniform in $[-2,2]$, so that the agent visits all parts of the phase plane from the first episodes, and the step size decreases with the episode number $n$ as $\alpha_n=0.5/(1+n/2000)$. The table starts at $Q=0$.

To measure what the agent can do we freeze its greedy policy $\arg\max_aQ(s,a)$ every thousand episodes and run it without exploration from 100 starting points near the hanging position, with $\theta$ and $\omega$ uniform in $[-0.3,0.3]$. A start counts as a swing-up when the pendulum is within 45 degrees of upright during the whole of the last ten of 40 time units. We train four agents with different random seeds in parallel.

The value-iteration reference solves the finite model with bilinear transition weights. The agent trains on continuous trajectories and observes their nearest grid points. These observations group continuous states with different successors, so the tabular process is an approximation. Policy agreement measures agreement with the finite-model reference. Swing-up success is measured separately on continuous dynamics. The 300-decision training episodes cut off a continuing task, and the final update retains its continuation term.

```python
QL_EPS, QL_A0, QL_ADEC, QL_STEPS = 0.1, 0.5, 2000.0, 300
N_BLOCKS, BLOCK = 30, 1000                      # 30 000 episodes, evaluated every 1000
N_SEEDS = 4
STARTS = jnp.array(np.random.default_rng(1).uniform(-0.3, 0.3, (100, 2)))

def ql_episode(carry, key):
    Q, n = carry
    alpha = QL_A0 / (1 + n / QL_ADEC)
    k1, k2, k3 = jax.random.split(key, 3)
    x0 = jnp.array([jax.random.uniform(k1, minval=-jnp.pi, maxval=jnp.pi), jax.random.uniform(k2, minval=-2, maxval=2)])
    def step(c, k):
        Q, x, G, disc = c
        s = cell(x); ka, kb = jax.random.split(k)
        a = jnp.where(jax.random.uniform(ka) < QL_EPS, jax.random.randint(kb, (), 0, 3), jnp.argmax(Q[s]))
        xn = propagate(x, ACTS[a]); r = -jnp.cos(xn[0]); sn = cell(xn)
        Q = Q.at[s, a].add(alpha * (r + GAMMA * Q[sn].max() - Q[s, a]))       # equation (8)
        return (Q, xn, G + disc * r, disc * GAMMA), None
    (Q, _, G, _), _ = lax.scan(step, (Q, x0, 0.0, 1.0), jax.random.split(k3, QL_STEPS))
    return (Q, n + 1), G

def success_rate(table):
    # fraction of the 100 starts near the bottom that end upright
    def one(x0):
        tr, _ = rollout_table(table, x0, N_EVAL)
        return jnp.mean(jnp.cos(tr[-100:, 0]) < -np.cos(np.pi / 4)) == 1.0
    return jax.vmap(one)(STARTS).mean()

@jax.jit
def ql_block(carry, key):
    carry, G = lax.scan(ql_episode, carry, jax.random.split(key, BLOCK))
    return carry, G, success_rate(carry[0].argmax(1))

ql_blocks = jax.jit(jax.vmap(ql_block))
os.makedirs("results", exist_ok=True)
t0 = time.time()
carry = (jnp.zeros((N_SEEDS, S, 3)), jnp.zeros(N_SEEDS))
keys = jax.random.split(jax.random.PRNGKey(56), N_SEEDS)
G_hist, succ_hist = [], []
for b in range(N_BLOCKS):
    keys, sub = jax.vmap(jax.random.split, out_axes=1)(keys)
    carry, G, succ = ql_blocks(carry, sub)
    G_hist.append(np.asarray(G)); succ_hist.append(np.asarray(succ))
    np.savez("results/lab056_training_checkpoint.npz", Q=np.asarray(carry[0]), episodes=np.asarray(carry[1]), keys=np.asarray(keys), G_hist=np.concatenate(G_hist, 1), succ_hist=np.stack(succ_hist, 1))
    print(f"Block {b + 1}/{N_BLOCKS}, elapsed {time.time()-t0:.1f} s, success {np.asarray(succ)}", flush=True)
G_hist = np.concatenate(G_hist, 1); succ_hist = np.stack(succ_hist, 1)
Q_learn = carry[0]
print(f"Q-learning, {N_SEEDS} seeds x {N_BLOCKS * BLOCK} episodes: {time.time() - t0:.0f} s")
print("success rate of the greedy policy after 5000, 10000, 20000, 30000 episodes, per seed:")
for b in (4, 9, 19, 29):
    print(f"  {(b + 1) * BLOCK:6d}: " + "  ".join(f"{v:.2f}" for v in succ_hist[:, b]))
print(f"optimal policy pi_*: success rate {float(success_rate(PI_STAR)):.2f}; greedy gamma = 0: {float(success_rate(PI_GREEDY)):.2f}")
```

## Summary

A pendulum with the dimensionless torque limit $u_{\max}=0.3$ reaches only $32.7$ degrees under a constant torque from rest, and the energy balance $dE/dt=\omega u-b\omega^2$ shows that the motor has to push along the motion to raise the energy to the separatrix. A grid of $51\times51$ points on the phase plane with bilinear interpolation weights as transition probabilities turns this swing-up into a finite Markov decision process with an explicit model. Value iteration solves its Bellman equation at the rate $\gamma$ per sweep, and the resulting policy pumps energy, with the torque along $\omega$ in $93.4\%$ of the grid points inside the separatrix, brakes near the top and balances the pendulum upright. From the hanging position it holds the pendulum upright from $t=19.4$ on, after five reversals of the motion. A greedy policy that maximises the immediate height never gets beyond $32.7$ degrees, which is the delayed reward of reinforcement learning in physical form. The smallest tested discount factor giving sustained swing-up is $\gamma=0.85$, a look-ahead of about $0.7$ time units, a fraction of one swing, consistent with each push paying off within the same half swing. Q-learning, which sees only the grid point of the state and the height after each action, reaches a greedy success rate of at least 95% from the fixed set of 100 starts for the first time after 5000 to 11 000 episodes in each of the four seeds, without any knowledge of the equation of motion. Its success rate fluctuates from one evaluation to the next. After the full 30 000 episodes the final success rates are 1.00, 0.19, 1.00 and 1.00, so a reliable result needs evaluation across seeds and training times.
